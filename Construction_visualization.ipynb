```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data=pd.read_csv("US_House_Price.csv")
```

```python
data['year'] = pd.to_datetime(data['DATE']).dt.year
data.drop(columns=['DATE','year'],inplace=True)
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(data.corr(),annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation")
```

The heatmap shows a strong and positive relationship between urban population, GDP and total houses.

```python
ax = sns.lineplot(x='urban_population', y='GDP', data = data)
ax.set_xlabel('Urban Population')
ax.set_ylabel('GDP')

ax.set_title('Relationship between Urban Population and GDP')
```

The line plot shows the relationship between Urban Population and GDP. Urban Population and GDP are strongly, positively correlated.

```python
ax = sns.scatterplot(x='building_permits', y='construction_unit', data = data,hue='house_for_sale_or_sold')
ax.set_xlabel('Building permits')
ax.set_ylabel('Construction Unit')
ax.set_title('Relationship between Building permits, Construction unit and house for sale')
plt.legend(title="House for sale or sold", bbox_to_anchor=(1.05, 1), loc='upper left')
```

The plot above illustrates that as building permits increase, the number of construction units and houses for sale or sold also increases in parallel.

```python
# Scatter plot
ax = sns.scatterplot(
    x='const_price_index', 
    y='home_price_index', 
    data=data, 
    hue='urban_population'  # correct column name
)

# Labels and title
ax.set_xlabel('Construction price index')
ax.set_ylabel('Home price index')
ax.set_title('Relationship between Construction Price Index, Home Price Index and Urban Population')

# Customize legend (optional)
plt.legend(title="Urban Population", bbox_to_anchor=(1.05, 1), loc='upper left')

plt.show()
```

The above plot shows that the Construction price index, Home price index, Urban population are strongly and positivily correlated.

```python
ax = sns.scatterplot(x='const_price_index', y='income', data = data,hue='total_houses')
ax.set_xlabel('Construction price index')
ax.set_ylabel('Income')
ax.set_title('Relationship between Construction Price Index, Income and Total Houses')
plt.legend(title="Total Housese", bbox_to_anchor=(1.05, 1), loc='upper left')
```

The above plot shows that the Construction price index, Income, Total Houses are strongly and positivily correlated.

```python
ax = sns.scatterplot(x='const_price_index', y='GDP', data = data,hue='housing_subsidies')
ax.set_xlabel('Construction price index')
ax.set_ylabel('GDP')
ax.set_title('Relationship between Construction Price Index, GDP and Housing Subsidies')
plt.legend(title="Housing Subsidies", bbox_to_anchor=(1.05, 1), loc='upper left')
```

The above plot shows that the Construction price index, GDP, Housing Subsidies are strongly and positivily correlated.

```python
ax = sns.scatterplot(x='delinquency_rate', y='unemployment_rate', data = data,hue='building_permits')
ax.set_xlabel('Delinquency Rate')
ax.set_ylabel('Unemployment Rate')
ax.set_title('Relationship between Delinquency Rate, Unemployment Rate and Building Permits')
plt.legend(title="Building Permits", bbox_to_anchor=(1.05, 1), loc='upper left')
```
